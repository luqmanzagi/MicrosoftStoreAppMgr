```python
import ujson as json
from pathlib import Path

result_dir = Path("results")
if not result_dir.is_dir():
    result_dir = Path("../results")

with (result_dir / "app_2026-10-20.json").open("r", encoding="utf-8") as f:
    results = json.load(f)

category_members = {}
for app in results:
    app_name = app.get("itemName")
    if not app_name:
        raise ValueError(f"Application is missing itemName: {app}")

    for category in app.get("categories") or ["Uncategorized"]:
        category_members.setdefault(category, []).append(app_name)

print(f"Total applications: {len(results)}")
```

```python
category_counts = {
    category: len(members)
    for category, members in sorted(category_members.items())
}
category_counts
```

```python
category_member_names = {
    category: sorted(members, key=str.casefold)
    for category, members in sorted(category_members.items())
}
category_member_names
```
