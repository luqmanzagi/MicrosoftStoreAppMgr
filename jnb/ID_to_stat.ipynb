```python
import json
from pathlib import Path

results_dir = Path("results")
if not results_dir.is_dir():
    results_dir = Path("../results")
with (results_dir / "manual_uninstall.txt").open(encoding="utf-8") as f:
    uninstall_ids = [line.strip() for line in f if line.strip()]

with (results_dir / "app_2026-09-21.json").open(encoding="utf-8") as f:
    apps = json.load(f)

apps_by_id = {
    str(app["itemID"]).casefold(): app
    for app in apps
    if app.get("itemID")
}
missing_ids = [item_id for item_id in uninstall_ids if item_id.casefold() not in apps_by_id]
if missing_ids:
    raise ValueError(f"IDs not found in app metadata: {missing_ids}")

matched_apps = [apps_by_id[item_id.casefold()] for item_id in uninstall_ids]
app_names = [app["itemName"] for app in matched_apps]
app_names
```

```python
from collections import Counter

category_counts = Counter(
    category
    for app in matched_apps
    for category in (app.get("categories") or ["Uncategorized"])
)
dict(sorted(category_counts.items()))
```
